```python
import pydicom
import numpy as np
import os
ruta_carpeta = "C:/Users/penak/OneDrive/Documents/SEMESTRE 8/PRUEBA_IMAGEN_KAROL/CATPHANKV"
dicoms = []
for i in os.listdir(ruta_carpeta): # listar todos los archivos de la carpeta
    nombre_archivo = os.fsdecode(i) 
    if nombre_archivo.endswith(".dcm"): # verificar que el archivo sea DICOM
        ruta_archivo = os.path.join(ruta_carpeta, nombre_archivo)  
        ds = pydicom.dcmread(ruta_archivo)
        if 'PixelData' in ds:
            dicoms.append(ds)
        else:
            print(f"Advertencia: El archivo {nombre_archivo} no contiene datos de píxeles.")
cortes_ordenados = sorted(dicoms, key=lambda x: float(getattr(x, 'SliceLocation', 0)))
cortes = cortes_ordenados
print(f"ds: {dicoms[0]}")
```
